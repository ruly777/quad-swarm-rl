```python
import torch
import numpy as np
from gym_art.quadrotor_multi.quadrotor_multi import QuadrotorEnvMulti
from sample_factory.algo.utils.make_env import make_env_func_batched
from sample_factory.utils.attr_dict import AttrDict

# Метод 1 - Прямое создание среды
def create_env(num_agents=1, use_numba=False, episode_duration=7):
    env = QuadrotorEnvMulti(
        num_agents=num_agents,
        use_numba=use_numba,
        episode_duration=episode_duration
    )
    return env

# Метод 2 - Инициализация в стиле Sample Factory
def create_env_sf():
    cfg = AttrDict({
        'env': 'quadrotor_multi',
        'quads_use_numba': True,
        'quads_num_agents': 1,
        'quads_mode': 'static_same_goal',
        'quads_episode_duration': 15.0,
        'num_envs_per_worker': 8
    })

    env_config = AttrDict(
        worker_index=0,
        vector_index=0,
        env_id=0,
    )

    env = make_env_func_batched(cfg, env_config)
    return env

# Тестируем оба метода
env1 = create_env()
env2 = create_env_sf()

obs1, info1 = env1.reset()
obs2, info2 = env2.reset()

print("Размерность наблюдений метод 1:", obs1.shape)
print("Размерность наблюдений метод 2:", obs2.shape)
```
